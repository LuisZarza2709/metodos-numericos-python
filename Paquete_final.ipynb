```python

import csv
import math
import matplotlib.pyplot as plt

ESTADO_MENU = "menu"
ESTADO_SALIR = "salir"
ESTADO_CAMBIAR = "cambiar"


# ============================================================
# FUNCIONES GENERALES
# ============================================================


def mostrar_integrantes():
    """Muestra la presentación general del paquete y los nombres de los integrantes."""
    print("=" * 70)
    print("PAQUETE FINAL - MÉTODOS NUMÉRICOS")
    print("-" * 70)
    print("Integrantes:")
    print("1. Williams Islas Hernández")
    print("2. Luis Alberto Avendaño Zarza")
    print("=" * 70)
    print()


def leer_real(mensaje):
    """Lee un número real desde teclado y repite la solicitud si la entrada no es válida."""
    while True:
        try:
            return float(input(mensaje))
        except ValueError:
            print("Error: entrada inválida. Ingrese un número real.")


def leer_entero(mensaje):
    """Lee un número entero desde teclado y repite la solicitud si la entrada no es válida."""
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("Error: entrada inválida. Ingrese un número entero.")


def leer_entero_positivo(mensaje):
    """Lee un entero positivo desde teclado."""
    while True:
        valor = leer_entero(mensaje)
        if valor <= 0:
            print("Error: debe ingresar un entero positivo.")
        else:
            return valor


def leer_entero_no_negativo(mensaje):
    """Lee un entero mayor o igual a cero desde teclado."""
    while True:
        valor = leer_entero(mensaje)
        if valor < 0:
            print("Error: debe ingresar un entero mayor o igual a cero.")
        else:
            return valor


def respuesta_si_no(mensaje):
    """Lee una respuesta afirmativa o negativa desde teclado."""
    while True:
        respuesta = input(mensaje).strip().upper()
        if respuesta in ("S", "SI", "SÍ"):
            return True
        if respuesta in ("N", "NO"):
            return False
        print("Respuesta inválida. Escriba S o N.")


def elegir_modo_desarrollo():
    """Permite elegir entre imprimir el desarrollo completo o solo los resultados principales."""
    while True:
        print("\nModo de impresión:")
        print("1. Desarrollo completo")
        print("2. Solo resultados principales")
        opcion = input("Seleccione una opción: ").strip()
        if opcion == "1":
            return True
        if opcion == "2":
            return False
        print("Opción no válida. Intente de nuevo.")


def imprimir_tabla_puntos(x, y):
    """Imprime una tabla de pares ordenados x, y."""
    print("\nTabla de valores:")
    print("{:<8} {:<15} {:<15}".format("Fila", "xi", "yi"))
    print("-" * 42)
    for i in range(len(x)):
        print("{:<8} {:<15.8f} {:<15.8f}".format(i, x[i], y[i]))
    print()


def ordenar_puntos(x, y):
    """Ordena dos listas de puntos tomando como referencia los valores de x."""
    pares = list(zip(x, y))
    pares.sort(key=lambda par: par[0])
    x_ordenado = [par[0] for par in pares]
    y_ordenado = [par[1] for par in pares]
    return x_ordenado, y_ordenado


def hay_x_repetidos(x, tolerancia=1e-12):
    """Verifica si una lista de valores x contiene elementos repetidos dentro de una tolerancia."""
    for i in range(len(x) - 1):
        if abs(x[i + 1] - x[i]) < tolerancia:
            return True
    return False


# ============================================================
# MÓDULO 1. MÉTODO DE BROYDEN
# ============================================================


def broyden_sistema_1(vector):
    """Evalúa uno de los sistemas no lineales disponibles para el método de Broyden."""
    x, y = vector
    f1 = x**2 + x*y + 2*y**2 - 5
    f2 = 5*y - 2*x*y**2 + 3
    return [f1, f2]


def broyden_sistema_2(vector):
    """Evalúa uno de los sistemas no lineales disponibles para el método de Broyden."""
    x, y = vector
    f1 = x**2 - 3*y**2 - 10
    f2 = 2*y**2 - 3*x*y + 1
    return [f1, f2]


def broyden_sistema_3(vector):
    """Evalúa uno de los sistemas no lineales disponibles para el método de Broyden."""
    x, y, z = vector
    f1 = 3*x**2 + y**2 - 8*y + 2*z**2 - 5
    f2 = -2*x**2 - 12*x + y**2 - 3*z**2 + 10
    f3 = -x + 2*y + 5*z
    return [f1, f2, f3]


def broyden_sistema_4(vector):
    """Evalúa uno de los sistemas no lineales disponibles para el método de Broyden."""
    x, y, z = vector
    f1 = x**2 + y**2 - 2*z + 3
    f2 = x + y + z - 5
    f3 = x**2 - y**2 + z**2 - 9
    return [f1, f2, f3]


def menu_sistemas_broyden():
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    while True:
        print("\nMenú de sistemas de ecuaciones")
        print("1. Sistema no lineal de 2 ecuaciones")
        print("   f1(x,y) = x^2 + xy + 2y^2 - 5")
        print("   f2(x,y) = 5y - 2xy^2 + 3")
        print()
        print("2. Sistema no lineal de 2 ecuaciones")
        print("   f1(x,y) = x^2 - 3y^2 - 10")
        print("   f2(x,y) = 2y^2 - 3xy + 1")
        print()
        print("3. Sistema no lineal de 3 ecuaciones")
        print("   f1(x,y,z) = 3x^2 + y^2 - 8y + 2z^2 - 5")
        print("   f2(x,y,z) = -2x^2 - 12x + y^2 - 3z^2 + 10")
        print("   f3(x,y,z) = -x + 2y + 5z")
        print()
        print("4. Sistema no lineal de 3 ecuaciones")
        print("   f1(x,y,z) = x^2 + y^2 - 2z + 3")
        print("   f2(x,y,z) = x + y + z - 5")
        print("   f3(x,y,z) = x^2 - y^2 + z^2 - 9")
        print()
        print("5. Regresar al menú principal")

        opcion = input("Seleccione una opción: ").strip()
        if opcion == "1":
            return broyden_sistema_1, 2, "Sistema 1"
        if opcion == "2":
            return broyden_sistema_2, 2, "Sistema 2"
        if opcion == "3":
            return broyden_sistema_3, 3, "Sistema 3"
        if opcion == "4":
            return broyden_sistema_4, 3, "Sistema 4"
        if opcion == "5":
            return None, None, None
        print("\nOpción no válida. Intente de nuevo.")


def imprimir_sistema_broyden(nombre_sistema, dimension):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    print("\nSistema elegido:")
    if nombre_sistema == "Sistema 1":
        print("f1(x,y) = x^2 + xy + 2y^2 - 5")
        print("f2(x,y) = 5y - 2xy^2 + 3")
    elif nombre_sistema == "Sistema 2":
        print("f1(x,y) = x^2 - 3y^2 - 10")
        print("f2(x,y) = 2y^2 - 3xy + 1")
    elif nombre_sistema == "Sistema 3":
        print("f1(x,y,z) = 3x^2 + y^2 - 8y + 2z^2 - 5")
        print("f2(x,y,z) = -2x^2 - 12x + y^2 - 3z^2 + 10")
        print("f3(x,y,z) = -x + 2y + 5z")
    elif nombre_sistema == "Sistema 4":
        print("f1(x,y,z) = x^2 + y^2 - 2z + 3")
        print("f2(x,y,z) = x + y + z - 5")
        print("f3(x,y,z) = x^2 - y^2 + z^2 - 9")
    print(f"Número de variables: {dimension}")
    print()


def obtener_valores_iniciales_broyden(dimension):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    print("\nLectura de valores iniciales")
    print("Ingrese los valores iniciales, la tolerancia y el máximo de iteraciones.")
    variables = ["x", "y", "z"]
    x0 = []
    for i in range(dimension):
        x0.append(leer_real(f"{variables[i]}0 = "))

    tolerancia = leer_real("Tolerancia = ")
    while tolerancia <= 0:
        print("Error: la tolerancia debe ser mayor que cero.")
        tolerancia = leer_real("Tolerancia = ")

    max_iteraciones = leer_entero_positivo("Máximo de iteraciones = ")
    return x0, tolerancia, max_iteraciones


def validar_datos_broyden(sistema, x0, tolerancia, max_iteraciones):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    errores = []
    if tolerancia <= 0:
        errores.append("La tolerancia debe ser mayor que cero.")
    if max_iteraciones <= 0:
        errores.append("El máximo de iteraciones debe ser positivo.")

    try:
        valores = sistema(x0)
        for valor in valores:
            if not math.isfinite(valor):
                errores.append("El sistema no tiene valores finitos con los datos iniciales.")
                break
    except (ValueError, ZeroDivisionError, OverflowError):
        errores.append("El sistema no se puede evaluar con los datos iniciales.")

    if errores:
        print("\nLos datos tienen los siguientes problemas:")
        for error in errores:
            print("-", error)
        print()
        return False

    print("\nLos datos son válidos para aplicar el método de Broyden.\n")
    return True


def broyden_copiar_vector(v):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    return [v[i] for i in range(len(v))]


def broyden_sumar_vectores(a, b):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    return [a[i] + b[i] for i in range(len(a))]


def broyden_restar_vectores(a, b):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    return [a[i] - b[i] for i in range(len(a))]


def broyden_multiplicar_vector_escalar(v, escalar):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    return [escalar * v[i] for i in range(len(v))]


def broyden_producto_punto(a, b):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    suma = 0.0
    for i in range(len(a)):
        suma += a[i] * b[i]
    return suma


def broyden_norma_vector(v):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    suma = 0.0
    for valor in v:
        suma += valor**2
    return math.sqrt(suma)


def broyden_multiplicar_matriz_vector(A, v):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    resultado = []
    for i in range(len(A)):
        suma = 0.0
        for j in range(len(v)):
            suma += A[i][j] * v[j]
        resultado.append(suma)
    return resultado


def broyden_sumar_matrices(A, B):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    resultado = []
    for i in range(len(A)):
        fila = []
        for j in range(len(A[0])):
            fila.append(A[i][j] + B[i][j])
        resultado.append(fila)
    return resultado


def broyden_resolver_sistema_lineal(A, b):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    n = len(A)
    matriz = []
    for i in range(n):
        fila = A[i][:]
        fila.append(b[i])
        matriz.append(fila)

    for col in range(n):
        fila_pivote = col
        for fila in range(col + 1, n):
            if abs(matriz[fila][col]) > abs(matriz[fila_pivote][col]):
                fila_pivote = fila

        if abs(matriz[fila_pivote][col]) < 1e-12:
            raise ValueError("No se puede resolver el sistema lineal porque aparece un pivote cero.")

        if fila_pivote != col:
            matriz[col], matriz[fila_pivote] = matriz[fila_pivote], matriz[col]

        pivote = matriz[col][col]
        for j in range(col, n + 1):
            matriz[col][j] = matriz[col][j] / pivote

        for fila in range(n):
            if fila != col:
                factor = matriz[fila][col]
                for j in range(col, n + 1):
                    matriz[fila][j] = matriz[fila][j] - factor * matriz[col][j]

    solucion = []
    for i in range(n):
        solucion.append(matriz[i][n])
    return solucion


def broyden_aproximar_jacobiano(sistema, x):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    n = len(x)
    h = 1e-6
    fx = sistema(x)
    J = [[0.0 for _ in range(n)] for _ in range(n)]

    for j in range(n):
        x_modificado = broyden_copiar_vector(x)
        x_modificado[j] += h
        fx_modificado = sistema(x_modificado)
        for i in range(n):
            J[i][j] = (fx_modificado[i] - fx[i]) / h

    return J


def broyden_actualizar_matriz(B, s, y):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    Bs = broyden_multiplicar_matriz_vector(B, s)
    y_menos_Bs = broyden_restar_vectores(y, Bs)
    denominador = broyden_producto_punto(s, s)

    if abs(denominador) < 1e-14:
        raise ValueError("No se puede actualizar la matriz porque el paso es demasiado pequeño.")

    n = len(B)
    correccion = [[0.0 for _ in range(n)] for _ in range(n)]
    for i in range(n):
        for j in range(n):
            correccion[i][j] = (y_menos_Bs[i] * s[j]) / denominador

    return broyden_sumar_matrices(B, correccion)


def broyden_metodo(sistema, x0, tolerancia, max_iteraciones):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    x_actual = broyden_copiar_vector(x0)
    B = broyden_aproximar_jacobiano(sistema, x_actual)
    iteraciones = []
    f_actual = sistema(x_actual)

    iteraciones.append({
        "iteracion": 0,
        "x": broyden_copiar_vector(x_actual),
        "norma_f": broyden_norma_vector(f_actual),
        "error": None
    })

    convergio = False
    error = None

    for k in range(1, max_iteraciones + 1):
        f_actual = sistema(x_actual)
        paso = broyden_resolver_sistema_lineal(B, broyden_multiplicar_vector_escalar(f_actual, -1.0))
        x_nuevo = broyden_sumar_vectores(x_actual, paso)
        f_nuevo = sistema(x_nuevo)
        error = broyden_norma_vector(broyden_restar_vectores(x_nuevo, x_actual))

        iteraciones.append({
            "iteracion": k,
            "x": broyden_copiar_vector(x_nuevo),
            "norma_f": broyden_norma_vector(f_nuevo),
            "error": error
        })

        if error < tolerancia or broyden_norma_vector(f_nuevo) < tolerancia:
            convergio = True
            x_actual = x_nuevo
            break

        s = broyden_restar_vectores(x_nuevo, x_actual)
        y = broyden_restar_vectores(f_nuevo, f_actual)
        B = broyden_actualizar_matriz(B, s, y)
        x_actual = x_nuevo

    return x_actual, error, convergio, iteraciones


def imprimir_tabla_iteraciones_broyden(iteraciones, dimension):
    """Imprime información formateada para el usuario."""
    variables = ["x", "y", "z"]
    print("\nSucesión de aproximaciones:")
    encabezado = "{:<10}".format("Iter")
    for i in range(dimension):
        encabezado += "{:>16}".format(variables[i])
    encabezado += "{:>18}{:>18}".format("||F(x)||", "Error")
    print(encabezado)
    print("-" * len(encabezado))

    for dato in iteraciones:
        fila = "{:<10}".format(dato["iteracion"])
        for valor in dato["x"]:
            fila += "{:>16.10f}".format(valor)
        fila += "{:>18.10e}".format(dato["norma_f"])
        if dato["error"] is None:
            fila += "{:>18}".format("-")
        else:
            fila += "{:>18.10e}".format(dato["error"])
        print(fila)
    print()


def imprimir_resultado_broyden(solucion, error, convergio, dimension):
    """Imprime información formateada para el usuario."""
    variables = ["x", "y", "z"]
    print("\n" + "=" * 70)
    print("RESULTADO DEL MÉTODO DE BROYDEN")
    print("=" * 70)

    if convergio:
        print("Solución aproximada:")
        for i in range(dimension):
            print(f"{variables[i]} = {solucion[i]:.10f}")
        if error is None:
            print("Error final = No disponible")
        else:
            print(f"Error final = {error:.10e}")
    else:
        print("No converge")
        print("No se alcanzó la tolerancia con el máximo de iteraciones indicado.")
        if error is not None:
            print(f"Último error calculado = {error:.10e}")
    print("=" * 70)
    print()


def resolver_broyden_con_datos(sistema, dimension, nombre_sistema, x0, tolerancia, max_iteraciones):
    """Apoya la lectura, el cálculo o la salida del módulo de Broyden."""
    print("\n" + "=" * 70)
    print("CÁLCULO POR MÉTODO DE BROYDEN")
    print("=" * 70)
    imprimir_sistema_broyden(nombre_sistema, dimension)
    print(f"Valores iniciales: {x0}")
    print(f"Tolerancia: {tolerancia}")
    print(f"Máximo de iteraciones: {max_iteraciones}")

    try:
        solucion, error, convergio, iteraciones = broyden_metodo(
            sistema, x0, tolerancia, max_iteraciones
        )
        imprimir_tabla_iteraciones_broyden(iteraciones, dimension)
        imprimir_resultado_broyden(solucion, error, convergio, dimension)
    except (ValueError, ZeroDivisionError, OverflowError) as error_calculo:
        print("\nNo converge")
        print("Ocurrió un problema durante el cálculo:")
        print(error_calculo)
        print()


def modulo_broyden():
    """Ejecuta el módulo interactivo del método de Broyden."""
    while True:
        print("\n" + "=" * 70)
        print("MÓDULO 1: SISTEMAS DE ECUACIONES NO LINEALES")
        print("=" * 70)
        print("1. Método de Broyden")
        print("2. Regresar al menú principal")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            sistema, dimension, nombre_sistema = menu_sistemas_broyden()
            if sistema is None:
                continue

            while True:
                x0, tolerancia, max_iteraciones = obtener_valores_iniciales_broyden(dimension)
                if not validar_datos_broyden(sistema, x0, tolerancia, max_iteraciones):
                    continue

                resolver_broyden_con_datos(
                    sistema, dimension, nombre_sistema, x0, tolerancia, max_iteraciones
                )

                print("\n¿Qué desea hacer ahora?")
                print("1. Probar otros datos iniciales con el mismo sistema")
                print("2. Elegir otro sistema")
                print("3. Regresar al menú principal")
                print("4. Salir del programa")
                siguiente = input("Seleccione una opción: ").strip()

                while siguiente not in ("1", "2", "3", "4"):
                    print("Opción no válida.")
                    siguiente = input("Seleccione una opción: ").strip()

                if siguiente == "1":
                    continue
                if siguiente == "2":
                    break
                if siguiente == "3":
                    print()
                    return ESTADO_MENU
                print("\nSaliendo del programa...")
                return ESTADO_SALIR

        elif opcion == "2":
            print()
            return ESTADO_MENU
        elif opcion == "3":
            print("\nSaliendo del programa...")
            return ESTADO_SALIR
        else:
            print("\nOpción no válida. Intente de nuevo.")


# ============================================================
# MÓDULO 2. INTERPOLACIÓN POLINOMIAL POR DIFERENCIAS DE NEWTON
# ============================================================


def newton_es_equidistante(x, tolerancia=1e-9):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    if len(x) < 2:
        return True, 0.0
    h = x[1] - x[0]
    for i in range(1, len(x) - 1):
        if abs((x[i + 1] - x[i]) - h) > tolerancia:
            return False, h
    return True, h


def newton_obtener_errores_tabla(x, y):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    errores = []
    if x is None or y is None:
        errores.append("La tabla no existe.")
        return errores
    if len(x) != len(y):
        errores.append("Las listas x e y no tienen la misma longitud.")
    if len(x) < 2:
        errores.append("Se requieren al menos 2 puntos.")

    ordenados = all(x[i] <= x[i + 1] for i in range(len(x) - 1))
    if not ordenados:
        errores.append("Los valores de x no están ordenados.")
    if hay_x_repetidos(x):
        errores.append("Hay valores de x repetidos.")
    if len(x) >= 2:
        equidistante, _ = newton_es_equidistante(x)
        if not equidistante:
            errores.append("La tabla no es equidistante.")
    return errores


def newton_construir_tabla_diferencias_finitas(y):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    n = len(y)
    tabla = [[0.0 for _ in range(n)] for _ in range(n)]
    for i in range(n):
        tabla[i][0] = y[i]
    for j in range(1, n):
        for i in range(n - j):
            tabla[i][j] = tabla[i + 1][j - 1] - tabla[i][j - 1]
    return tabla


def newton_imprimir_tabla_diferencias_finitas(x, tabla):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    n = len(x)
    encabezados = ["xi", "f(xi)"] + [f"Δ^{j}f" if j > 1 else "Δf" for j in range(1, n)]
    ancho = 14
    print("\nTabla de diferencias finitas:")
    for enc in encabezados:
        print(f"{enc:<{ancho}}", end="")
    print()
    print("-" * (ancho * len(encabezados)))
    for i in range(n):
        print(f"{x[i]:<{ancho}.8f}", end="")
        for j in range(n - i):
            print(f"{tabla[i][j]:<{ancho}.8f}", end="")
        print()
    print()


def newton_seleccionar_subtabla(x, y, xp, grado):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    m = grado + 1
    n = len(x)
    if m > n:
        return None, None, None

    indice_cercano = min(range(n), key=lambda i: abs(x[i] - xp))
    inicio = indice_cercano - grado // 2
    fin = inicio + m - 1

    if inicio < 0:
        inicio = 0
        fin = m - 1
    if fin >= n:
        fin = n - 1
        inicio = n - m

    x_sub = x[inicio:fin + 1]
    y_sub = y[inicio:fin + 1]
    return x_sub, y_sub, inicio


def newton_factorial(n):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado


def newton_evaluar_progresivo(x_sub, tabla, xp, grado, h):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    x0 = x_sub[0]
    s = (xp - x0) / h
    resultado = tabla[0][0]
    producto = 1.0
    terminos = []

    terminos.append({"orden": 0, "dato": tabla[0][0], "factor": "1", "divisor": 1, "valor": tabla[0][0]})

    for j in range(1, grado + 1):
        producto *= (s - (j - 1))
        valor = (producto / newton_factorial(j)) * tabla[0][j]
        resultado += valor
        if j == 1:
            factor = "s"
        else:
            factor = "s"
            for k in range(1, j):
                factor += f"(s-{k})"
        terminos.append({
            "orden": j,
            "dato": tabla[0][j],
            "factor": factor,
            "divisor": newton_factorial(j),
            "valor": valor
        })
    return resultado, s, terminos


def newton_evaluar_regresivo(x_sub, tabla, xp, grado, h):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    n = len(x_sub)
    xn = x_sub[-1]
    s = (xp - xn) / h
    resultado = tabla[n - 1][0]
    producto = 1.0
    terminos = []

    terminos.append({"orden": 0, "dato": tabla[n - 1][0], "factor": "1", "divisor": 1, "valor": tabla[n - 1][0]})

    for j in range(1, grado + 1):
        producto *= (s + (j - 1))
        dato = tabla[n - 1 - j][j]
        valor = (producto / newton_factorial(j)) * dato
        resultado += valor
        if j == 1:
            factor = "s"
        else:
            factor = "s"
            for k in range(1, j):
                factor += f"(s+{k})"
        terminos.append({
            "orden": j,
            "dato": dato,
            "factor": factor,
            "divisor": newton_factorial(j),
            "valor": valor
        })
    return resultado, s, terminos


def newton_elegir_metodo(x_sub, xp):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    distancia_inicio = abs(xp - x_sub[0])
    distancia_final = abs(x_sub[-1] - xp)
    if distancia_inicio <= distancia_final:
        return "progresivo"
    return "regresivo"


def newton_polinomio_texto_progresivo(x_sub, tabla, grado, h):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    partes = [f"{tabla[0][0]:.8f}"]
    for j in range(1, grado + 1):
        coef = tabla[0][j] / newton_factorial(j)
        termino = f"{coef:.8f}"
        termino += "(s)"
        for k in range(1, j):
            termino += f"(s-{k})"
        partes.append(termino)
    return " + ".join(partes), x_sub[0], h


def newton_polinomio_texto_regresivo(x_sub, tabla, grado, h):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    n = len(x_sub)
    partes = [f"{tabla[n - 1][0]:.8f}"]
    for j in range(1, grado + 1):
        coef = tabla[n - 1 - j][j] / newton_factorial(j)
        termino = f"{coef:.8f}"
        termino += "(s)"
        for k in range(1, j):
            termino += f"(s+{k})"
        partes.append(termino)
    return " + ".join(partes), x_sub[-1], h


def newton_imprimir_desarrollo_terminos(terminos):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    print("\nDesarrollo por términos:")
    print("{:<10} {:<18} {:<18} {:<12} {:<18}".format("Orden", "Dato", "Factor", "Divisor", "Valor"))
    print("-" * 76)
    for t in terminos:
        print("{:<10} {:<18.8f} {:<18} {:<12} {:<18.8f}".format(
            t["orden"], t["dato"], t["factor"], t["divisor"], t["valor"]
        ))
    print()


def newton_leer_tabla():
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    n = leer_entero_positivo("Ingrese el número de puntos en la tabla: ")
    while n < 2:
        print("Se requieren al menos 2 puntos.")
        n = leer_entero_positivo("Ingrese el número de puntos en la tabla: ")

    x = []
    y = []
    print("\nCapture los puntos (xi, yi):")
    for i in range(n):
        x.append(leer_real(f"x[{i}] = "))
        y.append(leer_real(f"y[{i}] = "))

    while True:
        imprimir_tabla_puntos(x, y)
        if respuesta_si_no("¿Son correctos los datos? (S/N): "):
            break

        fila = leer_entero_no_negativo("Ingrese la fila a corregir: ")
        while fila < 0 or fila >= n:
            print("Error: fila fuera de rango.")
            fila = leer_entero_no_negativo("Ingrese la fila a corregir: ")
        x[fila] = leer_real(f"Nuevo x[{fila}] = ")
        y[fila] = leer_real(f"Nuevo y[{fila}] = ")

    while True:
        ordenados = all(x[i] <= x[i + 1] for i in range(len(x) - 1))
        if not ordenados:
            print("\nLos datos no estaban ordenados. Se ordenarán automáticamente.")
            x, y = ordenar_puntos(x, y)

        imprimir_tabla_puntos(x, y)
        errores = newton_obtener_errores_tabla(x, y)

        if not errores:
            _, h = newton_es_equidistante(x)
            print(f"Los datos están ordenados y son equidistantes con h = {h:.8f}\n")
            return x, y

        print("La tabla tiene los siguientes problemas:")
        for error in errores:
            print("-", error)
        print()

        if not respuesta_si_no("¿Desea corregir una fila? (S/N): "):
            return None, None

        fila = leer_entero_no_negativo("Ingrese la fila que desea corregir: ")
        while fila < 0 or fila >= n:
            print("Error: fila fuera de rango.")
            fila = leer_entero_no_negativo("Ingrese la fila que desea corregir: ")
        x[fila] = leer_real(f"Nuevo x[{fila}] = ")
        y[fila] = leer_real(f"Nuevo y[{fila}] = ")


def newton_interpolar_con_tabla(x, y):
    """Apoya la interpolación por Diferencias de Newton con diferencias finitas."""
    while True:
        print("\n--- NUEVA INTERPOLACIÓN CON LA MISMA TABLA ---")
        print(f"Intervalo válido: [{x[0]:.8f}, {x[-1]:.8f}]")

        xp = leer_real("Ingrese el punto a interpolar: ")
        while xp < x[0] or xp > x[-1]:
            print("Error: el punto está fuera del intervalo [x0, xn].")
            xp = leer_real("Ingrese un punto dentro del intervalo: ")

        grado = leer_entero_no_negativo("Ingrese el grado del polinomio: ")
        while grado < 1 or grado >= len(x):
            if grado < 1:
                print("Error: el grado debe ser mayor o igual a 1.")
            else:
                print(f"Error: con {len(x)} puntos, el grado máximo permitido es {len(x)-1}.")
            grado = leer_entero_no_negativo("Ingrese otro grado: ")

        x_sub, y_sub, inicio = newton_seleccionar_subtabla(x, y, xp, grado)
        print("\nPuntos seleccionados para la interpolación:")
        imprimir_tabla_puntos(x_sub, y_sub)

        h = x_sub[1] - x_sub[0] if len(x_sub) > 1 else 0.0
        tabla = newton_construir_tabla_diferencias_finitas(y_sub)
        newton_imprimir_tabla_diferencias_finitas(x_sub, tabla)
        metodo = newton_elegir_metodo(x_sub, xp)

        if metodo == "progresivo":
            resultado, s, terminos = newton_evaluar_progresivo(x_sub, tabla, xp, grado, h)
            poli, base, h_texto = newton_polinomio_texto_progresivo(x_sub, tabla, grado, h)
            print("Método seleccionado: Newton progresivo")
            print(f"s = (x - x0)/h = (x - {base:.8f})/{h_texto:.8f}")
            print("Polinomio en función de s:")
            print(f"P(x) = {poli}")
        else:
            resultado, s, terminos = newton_evaluar_regresivo(x_sub, tabla, xp, grado, h)
            poli, base, h_texto = newton_polinomio_texto_regresivo(x_sub, tabla, grado, h)
            print("Método seleccionado: Newton regresivo")
            print(f"s = (x - xn)/h = (x - {base:.8f})/{h_texto:.8f}")
            print("Polinomio en función de s:")
            print(f"P(x) = {poli}")

        print(f"\nValor de s = {s:.8f}")
        newton_imprimir_desarrollo_terminos(terminos)
        print("Resultado de la interpolación:")
        print(f"f({xp:.8f}) ≈ {resultado:.10f}\n")

        print("¿Qué desea hacer ahora?")
        print("1. Interpolar otro punto con la misma tabla")
        print("2. Cambiar la tabla")
        print("3. Regresar al menú principal")
        print("4. Salir")
        opcion = input("Seleccione una opción: ").strip()
        while opcion not in ("1", "2", "3", "4"):
            print("Opción no válida.")
            opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            continue
        if opcion == "2":
            return ESTADO_CAMBIAR
        if opcion == "3":
            return ESTADO_MENU
        return ESTADO_SALIR


def modulo_interpolacion_newton():
    """Ejecuta el módulo interactivo de interpolación por Diferencias de Newton."""
    while True:
        print("\n" + "=" * 70)
        print("MÓDULO 2: INTERPOLACIÓN POLINOMIAL POR DIFERENCIAS DE NEWTON")
        print("=" * 70)
        print("1. Leer tabla e interpolar")
        print("2. Regresar al menú principal")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            x, y = newton_leer_tabla()
            if x is None or y is None:
                continue

            accion = newton_interpolar_con_tabla(x, y)
            if accion == ESTADO_CAMBIAR:
                continue
            if accion == ESTADO_MENU:
                print()
                return ESTADO_MENU
            if accion == ESTADO_SALIR:
                print("\nSaliendo del programa...")
                return ESTADO_SALIR

        elif opcion == "2":
            print()
            return ESTADO_MENU
        elif opcion == "3":
            print("\nSaliendo del programa...")
            return ESTADO_SALIR
        else:
            print("\nOpción no válida. Intente de nuevo.")


# ============================================================
# MÓDULO 3. SPLINE CÚBICO NATURAL
# ============================================================


def spline_obtener_errores_tabla(x, y):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    errores = []
    if x is None or y is None:
        errores.append("La tabla no existe.")
        return errores
    if len(x) != len(y):
        errores.append("Las listas x e y no tienen la misma longitud.")
    if len(x) < 3:
        errores.append("Se requieren al menos 3 puntos para construir un spline cúbico.")

    ordenados = all(x[i] <= x[i + 1] for i in range(len(x) - 1))
    if not ordenados:
        errores.append("Los valores de x no están ordenados.")
    if hay_x_repetidos(x):
        errores.append("Hay valores de x repetidos.")
    return errores


def spline_leer_tabla_manual():
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    n = leer_entero_positivo("Ingrese el número de puntos en la tabla: ")
    while n < 3:
        print("Se requieren al menos 3 puntos para construir un spline cúbico.")
        n = leer_entero_positivo("Ingrese el número de puntos en la tabla: ")

    x = []
    y = []
    print("\nCapture los puntos (xi, yi):")
    for i in range(n):
        x.append(leer_real(f"x[{i}] = "))
        y.append(leer_real(f"y[{i}] = "))

    while True:
        imprimir_tabla_puntos(x, y)
        if respuesta_si_no("¿Son correctos los datos? (S/N): "):
            break

        fila = leer_entero_no_negativo("Ingrese la fila a corregir: ")
        while fila < 0 or fila >= n:
            print("Error: fila fuera de rango.")
            fila = leer_entero_no_negativo("Ingrese la fila a corregir: ")
        x[fila] = leer_real(f"Nuevo x[{fila}] = ")
        y[fila] = leer_real(f"Nuevo y[{fila}] = ")

    return x, y


def spline_leer_tabla_csv():
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nLectura desde archivo CSV")
    print("El archivo debe tener dos columnas: x,y")
    print("Puede tener encabezado o no tenerlo.")
    print("Use punto decimal, por ejemplo: 2.65")
    ruta = input("Ingrese el nombre o ruta del archivo CSV: ").strip()

    if ruta == "":
        print("No se ingresó ningún archivo.")
        return None, None

    ruta = ruta.strip('"').strip("'")
    x = []
    y = []

    try:
        with open(ruta, "r", encoding="utf-8-sig", newline="") as archivo:
            muestra = archivo.read(1024)
            archivo.seek(0)
            try:
                dialecto = csv.Sniffer().sniff(muestra, delimiters=",;")
            except csv.Error:
                dialecto = csv.excel

            lector = csv.reader(archivo, dialecto)
            numero_fila = 0
            encabezado_revisado = False

            for fila in lector:
                numero_fila += 1
                if len(fila) == 0:
                    continue
                fila = [dato.strip() for dato in fila]
                if len(fila) < 2:
                    print(f"Error en la fila {numero_fila}: se esperaban dos columnas.")
                    return None, None

                if not encabezado_revisado:
                    encabezado_revisado = True
                    try:
                        xi = float(fila[0])
                        yi = float(fila[1])
                    except ValueError:
                        continue
                    x.append(xi)
                    y.append(yi)
                    continue

                try:
                    xi = float(fila[0])
                    yi = float(fila[1])
                except ValueError:
                    print(f"Error en la fila {numero_fila}: los datos deben ser numéricos.")
                    return None, None
                x.append(xi)
                y.append(yi)

    except FileNotFoundError:
        print("Error: no se encontró el archivo.")
        return None, None
    except PermissionError:
        print("Error: no se tiene permiso para abrir el archivo.")
        return None, None
    except OSError:
        print("Error: no se pudo abrir el archivo.")
        return None, None

    if len(x) == 0:
        print("El archivo no contiene datos válidos.")
        return None, None

    print("\nDatos leídos desde el archivo CSV:")
    imprimir_tabla_puntos(x, y)

    if respuesta_si_no("¿Desea corregir algún dato leído? (S/N): "):
        while True:
            fila = leer_entero_no_negativo("Ingrese la fila a corregir: ")
            while fila < 0 or fila >= len(x):
                print("Error: fila fuera de rango.")
                fila = leer_entero_no_negativo("Ingrese la fila a corregir: ")
            x[fila] = leer_real(f"Nuevo x[{fila}] = ")
            y[fila] = leer_real(f"Nuevo y[{fila}] = ")
            imprimir_tabla_puntos(x, y)
            if respuesta_si_no("¿Ya son correctos los datos? (S/N): "):
                break

    return x, y


def spline_cargar_ejemplo_prueba():
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    x = [2.65, 3.24, 4.19, 5.20, 5.95, 6.82, 8.56]
    y = [2.93, 3.98, 4.66, 4.90, 5.77, 6.14, 5.62]
    print("\nSe cargó un ejemplo de prueba.")
    imprimir_tabla_puntos(x, y)
    return x, y


def spline_obtener_tabla():
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    while True:
        print("\n" + "-" * 70)
        print("ENTRADA DE DATOS PARA SPLINE CÚBICO")
        print("-" * 70)
        print("1. Capturar datos manualmente")
        print("2. Cargar datos desde archivo CSV")
        print("3. Usar ejemplo de prueba")
        print("4. Regresar al menú principal")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            return spline_leer_tabla_manual()
        if opcion == "2":
            return spline_leer_tabla_csv()
        if opcion == "3":
            return spline_cargar_ejemplo_prueba()
        if opcion == "4":
            return None, None
        print("\nOpción no válida. Intente de nuevo.")


def spline_calcular_hi(x):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    h = []
    for i in range(len(x) - 1):
        h.append(x[i + 1] - x[i])
    return h


def spline_calcular_diferencias_divididas(x, y, h):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    diferencias = []
    for i in range(len(h)):
        diferencias.append((y[i + 1] - y[i]) / h[i])
    return diferencias


def spline_imprimir_calculos_hi(x, h):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nCálculo de los valores h_i:")
    print("h_i = x[i+1] - x[i]\n")
    for i in range(len(h)):
        print(f"h[{i}] = {x[i + 1]:.8f} - {x[i]:.8f} = {h[i]:.8f}")
    print()


def spline_imprimir_calculos_diferencias(x, y, h, diferencias):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nCálculo de diferencias divididas:")
    print("f[x_i, x_{i+1}] = (y[i+1] - y[i]) / h_i\n")
    for i in range(len(diferencias)):
        print(
            f"f[x{i}, x{i + 1}] = "
            f"({y[i + 1]:.8f} - {y[i]:.8f}) / {h[i]:.8f} "
            f"= {diferencias[i]:.8f}"
        )
    print()


def spline_imprimir_tabla(x, y, h, diferencias):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nTabla para spline cúbico:")
    print("{:<8} {:<15} {:<15} {:<15} {:<18}".format(
        "i", "xi", "yi", "hi", "f[xi,xi+1]"
    ))
    print("-" * 75)
    for i in range(len(x)):
        if i < len(h):
            print("{:<8} {:<15.8f} {:<15.8f} {:<15.8f} {:<18.8f}".format(
                i, x[i], y[i], h[i], diferencias[i]
            ))
        else:
            print("{:<8} {:<15.8f} {:<15.8f} {:<15} {:<18}".format(
                i, x[i], y[i], "-", "-"
            ))
    print()


def spline_imprimir_ecuaciones(h, diferencias):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    n = len(h) + 1
    print("\nConstrucción del sistema de ecuaciones:")
    print("Para spline cúbico natural se usa:")
    print("S0 = 0 y Sn = 0")
    print()
    print("h[i-1]S[i-1] + 2(h[i-1] + h[i])S[i] + h[i]S[i+1]")
    print("= 6(f[i] - f[i-1])")
    print()
    for i in range(1, n - 1):
        lado_derecho = 6 * (diferencias[i] - diferencias[i - 1])
        print(f"Para i = {i}:")
        print(
            f"{h[i - 1]:.8f}S{i - 1} + "
            f"2({h[i - 1]:.8f} + {h[i]:.8f})S{i} + "
            f"{h[i]:.8f}S{i + 1} "
            f"= 6({diferencias[i]:.8f} - {diferencias[i - 1]:.8f})"
        )
        print(
            f"{h[i - 1]:.8f}S{i - 1} + "
            f"{2 * (h[i - 1] + h[i]):.8f}S{i} + "
            f"{h[i]:.8f}S{i + 1} "
            f"= {lado_derecho:.8f}"
        )
        print()


def spline_construir_sistema(h, diferencias):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    n = len(h) + 1
    cantidad_incognitas = n - 2
    A = [[0.0 for _ in range(cantidad_incognitas)] for _ in range(cantidad_incognitas)]
    b = [0.0 for _ in range(cantidad_incognitas)]

    for i in range(1, n - 1):
        fila = i - 1
        if i - 1 >= 1:
            A[fila][fila - 1] = h[i - 1]
        A[fila][fila] = 2 * (h[i - 1] + h[i])
        if i + 1 <= n - 2:
            A[fila][fila + 1] = h[i]
        b[fila] = 6 * (diferencias[i] - diferencias[i - 1])
    return A, b


def spline_imprimir_matriz_sistema(A, b):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nSistema matricial:")
    print("A * S = b")
    print()
    cantidad_incognitas = len(A)
    encabezado = ""
    for i in range(1, cantidad_incognitas + 1):
        encabezado += f"{'S' + str(i):>14}"
    encabezado += f"{'b':>14}"
    print(encabezado)
    print("-" * (14 * (cantidad_incognitas + 1)))
    for i in range(cantidad_incognitas):
        for j in range(cantidad_incognitas):
            print(f"{A[i][j]:>14.8f}", end="")
        print(f"{b[i]:>14.8f}")
    print()


def spline_resolver_sistema_gauss(A, b):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    return broyden_resolver_sistema_lineal(A, b)


def spline_obtener_valores_S(A, b, cantidad_puntos):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    solucion_interna = spline_resolver_sistema_gauss(A, b)
    S = [0.0 for _ in range(cantidad_puntos)]
    for i in range(1, cantidad_puntos - 1):
        S[i] = solucion_interna[i - 1]
    return S


def spline_imprimir_valores_S(S):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nValores obtenidos de S_i:")
    print("{:<8} {:<15}".format("i", "S_i"))
    print("-" * 25)
    for i in range(len(S)):
        print("{:<8} {:<15.8f}".format(i, S[i]))
    print()


def spline_imprimir_formulas_coeficientes():
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nFórmulas para los coeficientes:")
    print("a_i = (S[i+1] - S[i]) / (6h_i)")
    print("b_i = S[i] / 2")
    print("c_i = f[x_i,x_{i+1}] - (h_i/6)(2S[i] + S[i+1])")
    print("d_i = y_i")
    print()


def spline_calcular_coeficientes(x, y, h, diferencias, S):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    coeficientes = []
    for i in range(len(h)):
        ai = (S[i + 1] - S[i]) / (6 * h[i])
        bi = S[i] / 2
        ci = diferencias[i] - (h[i] / 6) * (2 * S[i] + S[i + 1])
        di = y[i]
        coeficientes.append({
            "i": i,
            "a": ai,
            "b": bi,
            "c": ci,
            "d": di,
            "xi": x[i],
            "xf": x[i + 1]
        })
    return coeficientes


def spline_imprimir_calculos_coeficientes(x, y, h, diferencias, S):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nCálculo de coeficientes por intervalo:")
    for i in range(len(h)):
        ai = (S[i + 1] - S[i]) / (6 * h[i])
        bi = S[i] / 2
        ci = diferencias[i] - (h[i] / 6) * (2 * S[i] + S[i + 1])
        di = y[i]
        print(f"\nIntervalo {i}: [{x[i]:.8f}, {x[i + 1]:.8f}]")
        print(f"a[{i}] = ({S[i + 1]:.8f} - {S[i]:.8f}) / (6({h[i]:.8f})) = {ai:.8f}")
        print(f"b[{i}] = {S[i]:.8f} / 2 = {bi:.8f}")
        print(
            f"c[{i}] = {diferencias[i]:.8f} - "
            f"({h[i]:.8f}/6)(2({S[i]:.8f}) + {S[i + 1]:.8f}) "
            f"= {ci:.8f}"
        )
        print(f"d[{i}] = y[{i}] = {di:.8f}")
    print()


def spline_imprimir_tabla_coeficientes(coeficientes):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nTabla de coeficientes del spline cúbico:")
    print("{:<8} {:<15} {:<15} {:<15} {:<15} {:<20}".format(
        "i", "a_i", "b_i", "c_i", "d_i", "Intervalo"
    ))
    print("-" * 95)
    for dato in coeficientes:
        intervalo = f"[{dato['xi']:.8f}, {dato['xf']:.8f}]"
        print("{:<8} {:<15.8f} {:<15.8f} {:<15.8f} {:<15.8f} {:<20}".format(
            dato["i"], dato["a"], dato["b"], dato["c"], dato["d"], intervalo
        ))
    print()


def spline_texto_x_menos_xi(xi):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    if xi < 0:
        return f"(x + {abs(xi):.8f})"
    return f"(x - {xi:.8f})"


def spline_formatear_termino(coeficiente, texto_variable, es_primer_termino=False):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    if abs(coeficiente) < 1e-12:
        return ""
    if es_primer_termino:
        signo = "-" if coeficiente < 0 else ""
    else:
        signo = " - " if coeficiente < 0 else " + "
    return signo + f"{abs(coeficiente):.8f}{texto_variable}"


def spline_construir_polinomio_texto(dato):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    xi = dato["xi"]
    a = dato["a"]
    b = dato["b"]
    c = dato["c"]
    d = dato["d"]
    base = spline_texto_x_menos_xi(xi)
    partes = []

    termino_a = spline_formatear_termino(a, f"{base}^3", True)
    if termino_a != "":
        partes.append(termino_a)

    termino_b = spline_formatear_termino(b, f"{base}^2", len(partes) == 0)
    if termino_b != "":
        partes.append(termino_b)

    termino_c = spline_formatear_termino(c, f"{base}", len(partes) == 0)
    if termino_c != "":
        partes.append(termino_c)

    if len(partes) == 0:
        partes.append(f"{d:.8f}")
    else:
        if d < 0:
            partes.append(f" - {abs(d):.8f}")
        else:
            partes.append(f" + {d:.8f}")
    return "".join(partes)


def spline_imprimir_forma_general_polinomio():
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nForma general de cada polinomio:")
    print("g_i(x) = a_i(x - x_i)^3 + b_i(x - x_i)^2 + c_i(x - x_i) + d_i")
    print()


def spline_imprimir_polinomios(coeficientes):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\nPolinomios del spline cúbico natural:\n")
    for dato in coeficientes:
        i = dato["i"]
        xi = dato["xi"]
        xf = dato["xf"]
        polinomio = spline_construir_polinomio_texto(dato)
        print(f"g{i}(x) = {polinomio}")
        print(f"Intervalo: [{xi:.8f}, {xf:.8f}]")
        print()


def spline_evaluar_polinomio(dato, valor_x):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    xi = dato["xi"]
    return (
        dato["a"] * (valor_x - xi) ** 3
        + dato["b"] * (valor_x - xi) ** 2
        + dato["c"] * (valor_x - xi)
        + dato["d"]
    )


def spline_generar_puntos_intervalo(inicio, fin, cantidad=60):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    puntos = []
    if cantidad <= 1:
        return [inicio, fin]
    paso = (fin - inicio) / (cantidad - 1)
    for i in range(cantidad):
        puntos.append(inicio + i * paso)
    return puntos


def spline_graficar(x, y, coeficientes):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    try:
        import matplotlib.pyplot as plt
    except ImportError:
        print("\nNo se pudo cargar matplotlib.")
        print("Instale la librería con: pip install matplotlib")
        return

    valores_x = []
    valores_y = []
    for dato in coeficientes:
        xi = dato["xi"]
        xf = dato["xf"]
        puntos_intervalo = spline_generar_puntos_intervalo(xi, xf, 60)
        for valor_x in puntos_intervalo:
            valores_x.append(valor_x)
            valores_y.append(spline_evaluar_polinomio(dato, valor_x))

    plt.figure(figsize=(9, 6))
    plt.plot(valores_x, valores_y, label="Spline cúbico natural")
    plt.scatter(x, y, label="Puntos originales")
    for i in range(len(x)):
        plt.text(x[i], y[i], f"  P{i}")
    plt.title("Ajuste por spline cúbico natural")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)
    plt.legend()

    if respuesta_si_no("¿Desea guardar la gráfica como imagen? (S/N): "):
        nombre = input("Nombre del archivo, por ejemplo grafica_spline.png: ").strip()
        if nombre == "":
            nombre = "grafica_spline.png"
        if not nombre.lower().endswith((".png", ".jpg", ".jpeg", ".pdf")):
            nombre += ".png"
        plt.savefig(nombre, dpi=300, bbox_inches="tight")
        print(f"Gráfica guardada como: {nombre}")

    plt.show()


def spline_validar_tabla(x, y):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    while True:
        ordenados = all(x[i] <= x[i + 1] for i in range(len(x) - 1))
        if not ordenados:
            print("\nLos datos no estaban ordenados. Se ordenarán automáticamente.")
            x, y = ordenar_puntos(x, y)

        imprimir_tabla_puntos(x, y)
        errores = spline_obtener_errores_tabla(x, y)

        if not errores:
            print("La tabla es válida para construir el spline cúbico natural.\n")
            return x, y

        print("La tabla tiene los siguientes problemas:")
        for error in errores:
            print("-", error)
        print()

        if len(x) < 3:
            print("No se puede continuar con menos de 3 puntos.")
            return None, None

        if not respuesta_si_no("¿Desea corregir una fila? (S/N): "):
            return None, None

        fila = leer_entero_no_negativo("Ingrese la fila que desea corregir: ")
        while fila < 0 or fila >= len(x):
            print("Error: fila fuera de rango.")
            fila = leer_entero_no_negativo("Ingrese la fila que desea corregir: ")
        x[fila] = leer_real(f"Nuevo x[{fila}] = ")
        y[fila] = leer_real(f"Nuevo y[{fila}] = ")


def spline_resolver_con_tabla(x, y, desarrollo_completo):
    """Apoya la construcción, evaluación o presentación del spline cúbico natural."""
    print("\n" + "-" * 70)
    print("CÁLCULO DE h_i Y DIFERENCIAS DIVIDIDAS")
    print("-" * 70)
    h = spline_calcular_hi(x)
    diferencias = spline_calcular_diferencias_divididas(x, y, h)

    if desarrollo_completo:
        spline_imprimir_calculos_hi(x, h)
        spline_imprimir_calculos_diferencias(x, y, h, diferencias)
    spline_imprimir_tabla(x, y, h, diferencias)

    print("\n" + "-" * 70)
    print("SISTEMA PARA OBTENER LOS VALORES S_i")
    print("-" * 70)
    if desarrollo_completo:
        spline_imprimir_ecuaciones(h, diferencias)
    A, b = spline_construir_sistema(h, diferencias)
    spline_imprimir_matriz_sistema(A, b)

    try:
        S = spline_obtener_valores_S(A, b, len(x))
        spline_imprimir_valores_S(S)
    except ValueError as error:
        print("Error al resolver el sistema:")
        print(error)
        return

    print("\n" + "-" * 70)
    print("CÁLCULO DE COEFICIENTES a_i, b_i, c_i, d_i")
    print("-" * 70)
    if desarrollo_completo:
        spline_imprimir_formulas_coeficientes()
    coeficientes = spline_calcular_coeficientes(x, y, h, diferencias, S)
    if desarrollo_completo:
        spline_imprimir_calculos_coeficientes(x, y, h, diferencias, S)
    spline_imprimir_tabla_coeficientes(coeficientes)

    print("\n" + "-" * 70)
    print("POLINOMIOS DEL SPLINE CÚBICO NATURAL")
    print("-" * 70)
    spline_imprimir_forma_general_polinomio()
    spline_imprimir_polinomios(coeficientes)

    print("\n" + "-" * 70)
    print("GRÁFICA DEL SPLINE CÚBICO NATURAL")
    print("-" * 70)
    if respuesta_si_no("¿Desea mostrar la gráfica del spline? (S/N): "):
        spline_graficar(x, y, coeficientes)
    else:
        print("Se omitió la gráfica.")

    print("\nEl spline cúbico natural fue calculado correctamente.")


def modulo_spline_cubico():
    """Ejecuta el módulo interactivo de spline cúbico natural."""
    while True:
        print("\n" + "=" * 70)
        print("MÓDULO 3: AJUSTE POR SPLINE CÚBICO NATURAL")
        print("=" * 70)
        print("1. Resolver spline cúbico natural")
        print("2. Regresar al menú principal")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            while True:
                x, y = spline_obtener_tabla()
                if x is None or y is None:
                    break

                x, y = spline_validar_tabla(x, y)
                if x is None or y is None:
                    continue

                desarrollo_completo = elegir_modo_desarrollo()
                spline_resolver_con_tabla(x, y, desarrollo_completo)

                print("\n¿Qué desea hacer ahora?")
                print("1. Resolver otro spline")
                print("2. Regresar al menú principal")
                print("3. Salir")
                siguiente = input("Seleccione una opción: ").strip()
                while siguiente not in ("1", "2", "3"):
                    print("Opción no válida.")
                    siguiente = input("Seleccione una opción: ").strip()

                if siguiente == "1":
                    continue
                if siguiente == "2":
                    print()
                    return ESTADO_MENU
                print("\nSaliendo del programa...")
                return ESTADO_SALIR

        elif opcion == "2":
            print()
            return ESTADO_MENU
        elif opcion == "3":
            print("\nSaliendo del programa...")
            return ESTADO_SALIR
        else:
            print("\nOpción no válida. Intente de nuevo.")


# ============================================================
# MÓDULO 4. INTEGRACIÓN Y DERIVACIÓN NUMÉRICA
# ============================================================


def calculo_funcion_1(x):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    return x**4 * math.sqrt((3 + 2*x**2) / 3)


def calculo_funcion_2(x):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    return x**5 / ((x**2 + 4)**(1/5))


def calculo_funcion_3(x):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    return 5 * math.cos((x**3) / 2) + x**3 + math.log(x**2 + 1)


def calculo_menu_funciones():
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    while True:
        print("\nFunciones disponibles para integración y derivación")
        print("1. f(x) = x^4 * sqrt((3 + 2x^2)/3)")
        print("2. f(x) = x^5 / raíz quinta(x^2 + 4)")
        print("3. f(x) = 5cos(x^3/2) + x^3 + ln(x^2 + 1)")
        print("4. Regresar")
        opcion = input("Seleccione una opción: ").strip()
        if opcion == "1":
            return calculo_funcion_1, "x^4 * sqrt((3 + 2x^2)/3)"
        if opcion == "2":
            return calculo_funcion_2, "x^5 / raíz quinta(x^2 + 4)"
        if opcion == "3":
            return calculo_funcion_3, "5cos(x^3/2) + x^3 + ln(x^2 + 1)"
        if opcion == "4":
            return None, None
        print("\nOpción no válida. Intente de nuevo.")


def calculo_obtener_datos():
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    print("\nLectura de intervalo y particiones")
    a = leer_real("Límite inferior a = ")
    b = leer_real("Límite superior b = ")
    n = leer_entero_positivo("Número de particiones n = ")
    return a, b, n


def calculo_validar_datos(f, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    errores = []
    if a == b:
        errores.append("Los límites no pueden ser iguales.")
    if n <= 0:
        errores.append("El número de particiones debe ser positivo.")
    try:
        fa = f(a)
        fb = f(b)
        if not math.isfinite(fa) or not math.isfinite(fb):
            errores.append("La función no tiene valores finitos en los extremos.")
    except (ValueError, ZeroDivisionError, OverflowError):
        errores.append("La función no se puede evaluar correctamente en el intervalo.")

    if errores:
        print("\nLos datos tienen los siguientes problemas:")
        for error in errores:
            print("-", error)
        print()
        return False
    print("\nLos datos son válidos para integración o derivación numérica.\n")
    return True


def calculo_construir_tabla_funcion(f, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    h = (b - a) / n
    x = []
    y = []
    for i in range(n + 1):
        xi = a + i * h
        yi = f(xi)
        if not math.isfinite(yi):
            raise ValueError("La función no tiene valores finitos en uno de los puntos de la tabla.")
        x.append(xi)
        y.append(yi)
    return x, y, h


def calculo_imprimir_tabla_funcion(x, y):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    print("\nTabla de valores:")
    print("{:<8} {:<18} {:<18}".format("i", "xi", "f(xi)"))
    print("-" * 44)
    for i in range(len(x)):
        print("{:<8} {:<18.8f} {:<18.8f}".format(i, x[i], y[i]))
    print()


def calculo_regla_trapecio_compuesta(f, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    h = (b - a) / n
    suma = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        suma += f(a + i * h)
    return h * suma


def calculo_regla_simpson_un_tercio(f, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    if n % 2 != 0:
        raise ValueError("Simpson 1/3 requiere que n sea par.")
    h = (b - a) / n
    suma = f(a) + f(b)
    for i in range(1, n):
        xi = a + i * h
        if i % 2 == 0:
            suma += 2 * f(xi)
        else:
            suma += 4 * f(xi)
    return (h / 3) * suma


def calculo_regla_simpson_tres_octavos(f, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    if n % 3 != 0:
        raise ValueError("Simpson 3/8 requiere que n sea múltiplo de 3.")
    h = (b - a) / n
    suma = f(a) + f(b)
    for i in range(1, n):
        xi = a + i * h
        if i % 3 == 0:
            suma += 2 * f(xi)
        else:
            suma += 3 * f(xi)
    return (3 * h / 8) * suma


def calculo_elegir_metodo_integracion(n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    while True:
        print("\nMétodos disponibles para integración:")
        print("1. Regla del trapecio compuesta")
        print("2. Regla de Simpson 1/3")
        print("3. Regla de Simpson 3/8")
        print("4. Elegir automáticamente según n")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            return "trapecio"
        if opcion == "2":
            if n % 2 != 0:
                print("\nAviso: n no es par, por lo que no se puede usar Simpson 1/3.")
                continue
            return "simpson_13"
        if opcion == "3":
            if n % 3 != 0:
                print("\nAviso: n no es múltiplo de 3, por lo que no se puede usar Simpson 3/8.")
                continue
            return "simpson_38"
        if opcion == "4":
            if n % 3 == 0:
                return "simpson_38"
            if n % 2 == 0:
                return "simpson_13"
            return "trapecio"
        print("Opción no válida.")


def calculo_resolver_integracion_con_datos(f, nombre_funcion, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    print("\n" + "=" * 70)
    print("CÁLCULO DE INTEGRACIÓN NUMÉRICA")
    print("=" * 70)
    print(f"Función: f(x) = {nombre_funcion}")
    print(f"Intervalo: [{a:.8f}, {b:.8f}]")
    print(f"Particiones: n = {n}")
    h = (b - a) / n
    print(f"h = (b - a)/n = {h:.8f}")

    try:
        if respuesta_si_no("¿Desea mostrar la tabla de valores? (S/N): "):
            x, y, _ = calculo_construir_tabla_funcion(f, a, b, n)
            calculo_imprimir_tabla_funcion(x, y)

        metodo = calculo_elegir_metodo_integracion(n)
        if metodo == "trapecio":
            resultado = calculo_regla_trapecio_compuesta(f, a, b, n)
            nombre_metodo = "Regla del trapecio compuesta"
        elif metodo == "simpson_13":
            resultado = calculo_regla_simpson_un_tercio(f, a, b, n)
            nombre_metodo = "Regla de Simpson 1/3"
        else:
            resultado = calculo_regla_simpson_tres_octavos(f, a, b, n)
            nombre_metodo = "Regla de Simpson 3/8"

        print("\nResultado de la integración:")
        print(f"Método empleado: {nombre_metodo}")
        print(f"Integral aproximada = {resultado:.10f}")
        print()
    except (ValueError, ZeroDivisionError, OverflowError) as error:
        print("\nOcurrió un error durante la integración:")
        print(error)
        print()


def calculo_calcular_derivadas_numericas(f, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    if n < 2:
        raise ValueError("Para derivación numérica se requieren al menos 2 particiones.")

    x, y, h = calculo_construir_tabla_funcion(f, a, b, n)
    derivadas = []
    metodos = []
    for i in range(n + 1):
        if i == 0:
            derivada = (-3*y[i] + 4*y[i + 1] - y[i + 2]) / (2*h)
            metodo = "Adelante"
        elif i == n:
            derivada = (3*y[i] - 4*y[i - 1] + y[i - 2]) / (2*h)
            metodo = "Atrás"
        else:
            derivada = (y[i + 1] - y[i - 1]) / (2*h)
            metodo = "Centrada"
        derivadas.append(derivada)
        metodos.append(metodo)
    return x, y, derivadas, metodos, h


def calculo_imprimir_tabla_derivacion(x, y, derivadas, metodos):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    print("\nTabla de derivación numérica:")
    print("{:<8} {:<15} {:<15} {:<18} {:<15}".format(
        "i", "xi", "f(xi)", "f'(xi)", "Método"
    ))
    print("-" * 75)
    for i in range(len(x)):
        print("{:<8} {:<15.8f} {:<15.8f} {:<18.8f} {:<15}".format(
            i, x[i], y[i], derivadas[i], metodos[i]
        ))
    print()


def calculo_resolver_derivacion_con_datos(f, nombre_funcion, a, b, n):
    """Apoya el módulo obligatorio de integración y derivación numérica."""
    print("\n" + "=" * 70)
    print("CÁLCULO DE DERIVACIÓN NUMÉRICA")
    print("=" * 70)
    print(f"Función: f(x) = {nombre_funcion}")
    print(f"Intervalo: [{a:.8f}, {b:.8f}]")
    print(f"Particiones: n = {n}")
    try:
        x, y, derivadas, metodos, h = calculo_calcular_derivadas_numericas(f, a, b, n)
        print(f"h = (b - a)/n = {h:.8f}")
        print("Métodos empleados:")
        print("- Extremo izquierdo: fórmula hacia adelante de tres puntos")
        print("- Puntos interiores: fórmula centrada")
        print("- Extremo derecho: fórmula hacia atrás de tres puntos")
        calculo_imprimir_tabla_derivacion(x, y, derivadas, metodos)
    except (ValueError, ZeroDivisionError, OverflowError) as error:
        print("\nOcurrió un error durante la derivación:")
        print(error)
        print()


def modulo_integracion_derivacion():
    """Ejecuta el módulo interactivo de integración y derivación numérica."""
    funcion = None
    nombre_funcion = None
    a = None
    b = None
    n = None

    while True:
        print("\n" + "=" * 70)
        print("MÓDULO 4: INTEGRACIÓN Y DERIVACIÓN")
        print("=" * 70)
        print("1. Elegir la función")
        print("2. Leer intervalo y particiones")
        print("3. Integración")
        print("4. Derivación numérica")
        print("5. Regresar al menú principal")
        print("6. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            funcion, nombre_funcion = calculo_menu_funciones()
            if funcion is not None:
                print(f"\nFunción seleccionada: f(x) = {nombre_funcion}\n")
        elif opcion == "2":
            if funcion is None:
                print("\nPrimero debe elegir una función con la opción 1.\n")
            else:
                a, b, n = calculo_obtener_datos()
                if not calculo_validar_datos(funcion, a, b, n):
                    a = None
                    b = None
                    n = None
        elif opcion == "3":
            if funcion is None:
                print("\nPrimero debe elegir una función con la opción 1.\n")
            elif a is None or b is None or n is None:
                print("\nPrimero debe leer intervalo y particiones con la opción 2.\n")
            else:
                calculo_resolver_integracion_con_datos(funcion, nombre_funcion, a, b, n)
        elif opcion == "4":
            if funcion is None:
                print("\nPrimero debe elegir una función con la opción 1.\n")
            elif a is None or b is None or n is None:
                print("\nPrimero debe leer intervalo y particiones con la opción 2.\n")
            else:
                calculo_resolver_derivacion_con_datos(funcion, nombre_funcion, a, b, n)
        elif opcion == "5":
            print()
            return ESTADO_MENU
        elif opcion == "6":
            print("\nSaliendo del programa...")
            return ESTADO_SALIR
        else:
            print("\nOpción no válida. Intente de nuevo.")


# ============================================================
# MÓDULO EXTRA. ROMBERG
# ============================================================

MAX_NIVELES_ROMBERG = 15
MAX_DIGITOS_ROMBERG = 10


def romberg_funcion_1(x):
    """Apoya el módulo extra de integración por Romberg."""
    return x**4 * math.sqrt((3 + 2*x**2) / 3)


def romberg_funcion_2(x):
    """Apoya el módulo extra de integración por Romberg."""
    return x**5 / ((x**2 + 4)**(1/5))


def menu_funciones_romberg():
    """Apoya el módulo extra de integración por Romberg."""
    while True:
        print("\nFunciones disponibles para Romberg")
        print("1. f(x) = x^4 * sqrt((3 + 2x^2)/3)")
        print("2. f(x) = x^5 / raíz quinta(x^2 + 4)")
        print("3. Regresar")
        opcion = input("Seleccione una opción: ").strip()
        if opcion == "1":
            return romberg_funcion_1, "x^4 * sqrt((3 + 2x^2)/3)"
        if opcion == "2":
            return romberg_funcion_2, "x^5 / raíz quinta(x^2 + 4)"
        if opcion == "3":
            return None, None
        print("\nOpción no válida. Intente de nuevo.")


def validar_digitos_romberg(digitos):
    """Apoya el módulo extra de integración por Romberg."""
    if digitos > MAX_DIGITOS_ROMBERG:
        print(f"\nAviso: por precisión de Python se usará un máximo de {MAX_DIGITOS_ROMBERG} dígitos.")
        return MAX_DIGITOS_ROMBERG
    return digitos


def calcular_tolerancia_romberg(digitos):
    """Apoya el módulo extra de integración por Romberg."""
    return 0.5 * 10**(-digitos)


def obtener_datos_romberg():
    """Apoya el módulo extra de integración por Romberg."""
    print("\nLectura de datos para Romberg")
    a = leer_real("Límite inferior a = ")
    b = leer_real("Límite superior b = ")
    digitos = leer_entero_positivo("Dígitos de precisión = ")
    digitos = validar_digitos_romberg(digitos)
    return a, b, digitos


def validar_datos_romberg(f, a, b):
    """Apoya el módulo extra de integración por Romberg."""
    errores = []
    if a == b:
        errores.append("Los límites de integración no pueden ser iguales.")
    try:
        fa = f(a)
        fb = f(b)
        if not math.isfinite(fa) or not math.isfinite(fb):
            errores.append("La función no tiene valores finitos en los extremos.")
    except (ValueError, ZeroDivisionError, OverflowError):
        errores.append("La función no se puede evaluar correctamente en el intervalo.")

    if errores:
        print("\nLos datos tienen los siguientes problemas:")
        for error in errores:
            print("-", error)
        print()
        return False
    print("\nLos datos son válidos para aplicar Romberg.\n")
    return True


def calcular_trapecio_inicial_romberg(f, a, b):
    """Apoya el módulo extra de integración por Romberg."""
    h = b - a
    I0 = (h / 2) * (f(a) + f(b))
    return I0, h


def calcular_siguiente_trapecio_romberg(f, a, b, J, I_anterior):
    """Apoya el módulo extra de integración por Romberg."""
    n = 2**J
    h = (b - a) / n
    suma_nueva = 0.0
    puntos_nuevos = []

    for k in range(1, 2**(J - 1) + 1):
        x_nuevo = a + (2*k - 1) * h
        fx_nuevo = f(x_nuevo)
        if not math.isfinite(fx_nuevo):
            raise ValueError("La función no tiene valores finitos en uno de los puntos evaluados.")
        suma_nueva += fx_nuevo
        puntos_nuevos.append((x_nuevo, fx_nuevo))

    I_nuevo = (I_anterior / 2) + h * suma_nueva
    return I_nuevo, h, n, suma_nueva, puntos_nuevos


def construir_romberg_hasta_precision(f, a, b, tolerancia):
    """Apoya el módulo extra de integración por Romberg."""
    R = []
    datos_trapecio = []
    I0, h0 = calcular_trapecio_inicial_romberg(f, a, b)
    R.append([I0])
    datos_trapecio.append({"J": 0, "n": 1, "h": h0, "I": I0, "suma_nueva": 0.0, "puntos_nuevos": []})

    error_aproximado = None
    alcanzo_precision = False

    for J in range(1, MAX_NIVELES_ROMBERG):
        I_nuevo, h, n, suma_nueva, puntos_nuevos = calcular_siguiente_trapecio_romberg(
            f, a, b, J, R[J - 1][0]
        )
        fila = [I_nuevo]
        for k in range(1, J + 1):
            valor = fila[k - 1] + (fila[k - 1] - R[J - 1][k - 1]) / (4**k - 1)
            fila.append(valor)

        R.append(fila)
        datos_trapecio.append({
            "J": J,
            "n": n,
            "h": h,
            "I": I_nuevo,
            "suma_nueva": suma_nueva,
            "puntos_nuevos": puntos_nuevos
        })

        error_aproximado = abs(R[J][J] - R[J - 1][J - 1])
        if error_aproximado < tolerancia:
            alcanzo_precision = True
            break

    return R, datos_trapecio, error_aproximado, alcanzo_precision


def imprimir_desarrollo_trapecio_romberg(datos_trapecio):
    """Apoya el módulo extra de integración por Romberg."""
    print("\n" + "-" * 70)
    print("DESARROLLO DEL TRAPECIO RECURSIVO")
    print("-" * 70)
    print("I(J) = I(J-1)/2 + h * suma de nuevas evaluaciones")
    print()

    for dato in datos_trapecio:
        J = dato["J"]
        print(f"J = {J}")
        print(f"n = {dato['n']}")
        print(f"h = {dato['h']:.10f}")
        if J == 0:
            print(f"I(0) = h/2 [f(a) + f(b)] = {dato['I']:.10f}")
        else:
            print("Puntos nuevos evaluados:")
            for x_nuevo, fx_nuevo in dato["puntos_nuevos"]:
                print(f"x = {x_nuevo:.10f}, f(x) = {fx_nuevo:.10f}")
            print(f"Suma nueva = {dato['suma_nueva']:.10f}")
            print(f"I({J}) = I({J - 1})/2 + h(suma nueva) = {dato['I']:.10f}")
        print()


def imprimir_tabla_trapecio_romberg(datos_trapecio):
    """Apoya el módulo extra de integración por Romberg."""
    print("\nTabla de aproximaciones por trapecio recursivo:")
    print("{:<8} {:<12} {:<15} {:<18}".format("J", "n", "h", "I(J)"))
    print("-" * 55)
    for dato in datos_trapecio:
        print("{:<8} {:<12} {:<15.10f} {:<18.10f}".format(
            dato["J"], dato["n"], dato["h"], dato["I"]
        ))
    print()


def imprimir_tabla_romberg(R):
    """Apoya el módulo extra de integración por Romberg."""
    print("\nTabla de Romberg:")
    columnas = len(R)
    encabezado = "{:<8}".format("J")
    for j in range(columnas):
        encabezado += "{:>18}".format(f"R{j}")
    print(encabezado)
    print("-" * (8 + 18 * columnas))

    for i in range(len(R)):
        fila = "{:<8}".format(i)
        for j in range(columnas):
            if j <= i:
                fila += "{:>18.10f}".format(R[i][j])
            else:
                fila += "{:>18}".format("")
        print(fila)
    print()


def imprimir_resultado_romberg(R, error_aproximado, alcanzo_precision, tolerancia, digitos):
    """Apoya el módulo extra de integración por Romberg."""
    ultimo_J = len(R) - 1
    resultado = R[ultimo_J][ultimo_J]
    print("\n" + "=" * 70)
    print("RESULTADO DE ROMBERG")
    print("=" * 70)
    print(f"Integral aproximada = {resultado:.10f}")
    print(f"Dígitos de precisión solicitados = {digitos}")
    print(f"Tolerancia usada = {tolerancia:.10e}")
    print(f"Niveles utilizados = {ultimo_J + 1}")
    if error_aproximado is None:
        print("Error aproximado = No disponible")
    else:
        print(f"Error aproximado = {error_aproximado:.10e}")
    if alcanzo_precision:
        print("Estado: se alcanzó la precisión solicitada.")
    else:
        print("Estado: no se alcanzó la precisión con el máximo de niveles permitido.")
    print("=" * 70)
    print()


def resolver_romberg_con_datos(f, nombre_funcion, a, b, digitos):
    """Apoya el módulo extra de integración por Romberg."""
    tolerancia = calcular_tolerancia_romberg(digitos)
    print("\n" + "=" * 70)
    print("CÁLCULO EXTRA POR ROMBERG")
    print("=" * 70)
    print(f"Función: f(x) = {nombre_funcion}")
    print(f"Intervalo: [{a:.10f}, {b:.10f}]")
    print(f"Dígitos de precisión: {digitos}")
    print(f"Tolerancia: {tolerancia:.10e}")
    desarrollo_completo = elegir_modo_desarrollo()

    try:
        R, datos_trapecio, error_aproximado, alcanzo_precision = construir_romberg_hasta_precision(
            f, a, b, tolerancia
        )
        if desarrollo_completo:
            imprimir_desarrollo_trapecio_romberg(datos_trapecio)
        imprimir_tabla_trapecio_romberg(datos_trapecio)
        imprimir_tabla_romberg(R)
        imprimir_resultado_romberg(R, error_aproximado, alcanzo_precision, tolerancia, digitos)
    except (ValueError, ZeroDivisionError, OverflowError) as error:
        print("\nOcurrió un error durante el cálculo de Romberg:")
        print(error)
        print()


def modulo_romberg_extra():
    """Ejecuta el módulo extra de integración por Romberg."""
    while True:
        print("\n" + "=" * 70)
        print("MÓDULO EXTRA: ROMBERG")
        print("=" * 70)
        print("1. Integración por Romberg")
        print("2. Regresar al menú principal")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            while True:
                f, nombre_funcion = menu_funciones_romberg()
                if f is None:
                    break

                a, b, digitos = obtener_datos_romberg()
                if not validar_datos_romberg(f, a, b):
                    continue

                resolver_romberg_con_datos(f, nombre_funcion, a, b, digitos)

                print("\n¿Qué desea hacer ahora?")
                print("1. Realizar otro cálculo de Romberg")
                print("2. Elegir otra función")
                print("3. Regresar al menú principal")
                print("4. Salir")
                siguiente = input("Seleccione una opción: ").strip()
                while siguiente not in ("1", "2", "3", "4"):
                    print("Opción no válida.")
                    siguiente = input("Seleccione una opción: ").strip()

                if siguiente == "1":
                    continue
                if siguiente == "2":
                    break
                if siguiente == "3":
                    print()
                    return ESTADO_MENU
                print("\nSaliendo del programa...")
                return ESTADO_SALIR

        elif opcion == "2":
            print()
            return ESTADO_MENU
        elif opcion == "3":
            print("\nSaliendo del programa...")
            return ESTADO_SALIR
        else:
            print("\nOpción no válida. Intente de nuevo.")


# ============================================================
# MENÚ PRINCIPAL GENERAL
# ============================================================


def menu_principal():
    """Controla la navegación principal del paquete final."""
    while True:
        mostrar_integrantes()
        print("MENÚ PRINCIPAL")
        print("1. Sistemas de ecuaciones no lineales - Método de Broyden")
        print("2. Interpolación polinomial por Diferencias de Newton")
        print("3. Ajuste por Spline Cúbico Natural")
        print("4. Integración y derivación")
        print("5. Extra: Romberg")
        print("6. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            estado = modulo_broyden()
            if estado == ESTADO_SALIR:
                break
        elif opcion == "2":
            estado = modulo_interpolacion_newton()
            if estado == ESTADO_SALIR:
                break
        elif opcion == "3":
            estado = modulo_spline_cubico()
            if estado == ESTADO_SALIR:
                break
        elif opcion == "4":
            estado = modulo_integracion_derivacion()
            if estado == ESTADO_SALIR:
                break
        elif opcion == "5":
            estado = modulo_romberg_extra()
            if estado == ESTADO_SALIR:
                break
        elif opcion == "6":
            print("\nSaliendo del programa...")
            break
        else:
            print("\nOpción no válida. Intente de nuevo.\n")



if __name__ == "__main__":
    menu_principal()


    #Pruebas

# 1. Entrar a Broyden y probar al menos un sistema de 2 variables.
# 2. Entrar a Newton con una tabla equidistante y revisar el polinomio en s.
# 3. Entrar a Spline con captura manual o ejemplo de prueba.
# 4. Entrar a Integración y derivación; probar Simpson y derivación.
# 5. Entrar a Romberg y revisar tabla de trapecio y tabla de Romberg.
# 6. Probar regresar al menú principal desde todos los módulos.
# 7. Probar la opción Salir desde el menú principal y desde módulos.
```
