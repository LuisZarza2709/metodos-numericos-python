```python
import math

# Definición de funciones
def f1(x):
    return x**4 * math.sqrt((3 + 2*x**2)/3)

def f2(x):
    return x**5 / math.sqrt(x**2 + 4)

def f3(x):
    return (x**2 + math.sin(x)) / (x - 2) + 3

def menu_funciones():
    print("Seleccione la función a integrar:")
    print("1. f(x) = x^4 * sqrt((3 + 2x^2)/3)")
    print("2. f(x) = x^5 / sqrt(x^2 + 4)")
    print("3. f(x) = (x^2 + sin(x)) / (x - 2) + 3")
    opcion = input("Ingrese opción (1/2/3): ")
    if opcion == "1":
        return f1
    elif opcion == "2":
        return f2
    elif opcion == "3":
        return f3
    else:
        print("Opción inválida, seleccionando f1 por defecto.")
        return f1

# Regla del trapecio
def trapecio(f, a, b, n):
    h = (b - a) / n
    suma = 0.5 * (f(a) + f(b))
    for i in range(1, n):
        suma += f(a + i*h)
    return suma * h

# Extrapolación de Romberg
def romberg(f, a, b, n):
    R = [[0 for _ in range(n)] for _ in range(n)]
    for i in range(n):
        ni = 2**i
        R[i][0] = trapecio(f, a, b, ni)
    for j in range(1, n):
        for i in range(j, n):
            R[i][j] = R[i][j-1] + (R[i][j-1] - R[i-1][j-1]) / (4**j - 1)
    return R[n-1][n-1]

# Función principal
def main():
    while True:
        f = menu_funciones()
        a = float(input("Ingrese límite inferior a: "))
        b = float(input("Ingrese límite superior b: "))
        n = int(input("Ingrese número de subintervalos inicial n: "))

        # Trapecio
        trap = trapecio(f, a, b, n)
        print(f"Resultado con regla del trapecio: {trap:.6f}")

        # Romberg con extrapolación hasta n pasos
        R = romberg(f, a, b, int(math.log2(n)))
        print(f"Resultado con extrapolación de Romberg: {R:.6f}")

        continuar = input("¿Desea integrar otra función? (s/n): ")
        if continuar.lower() != 's':
            break

if __name__ == "__main__":
    main()
```
